### Include necessary libraries

```python
import pandas as pd
import numpy as np
import scipy.stats
import matplotlib.pyplot as plt
```

### Ex.01 - Plot histogram

```python
data = pd.read_excel('Ex3_hints_sample.xlsx',index_col = 'Index')
print(data)

x = np.arange(0, 30, step = 1)
data.hist(column = 'Price', bins = x, density = True, alpha = 0.4)

plt.show()
```

### Ex.02 - Mean and Standard deviation

```python
print('Mean = ', data['Price'].mean())
print(' ')
print('STD = ', data['Price'].std())
```

### Ex.03 - Plot pdf

```python
x = np.arange(-5, 5, step = 1e-2)
n_pdf = scipy.stats.norm.pdf(x, loc = 0, scale = 1) 

plt.plot(x, n_pdf)
plt.grid()
plt.xlabel('x')
plt.ylabel('pdf')
plt.show()
```

### Ex.04 - Plot two pdfs

```python
x = np.arange(-10, 30, step = 5e-2)

n_pdf1 = scipy.stats.norm.pdf(x, loc = 2, scale = 2) 
n_pdf2 = scipy.stats.norm.pdf(x, loc = 22, scale = 2) 

fig, ax = plt.subplots()

ax.plot(x, n_pdf1)
ax.plot(x, n_pdf2)

plt.grid()
plt.xlabel('x')
plt.ylabel('pdf')
plt.show()
```

### Ex.05 - groupby

```python
a = data.groupby('Class')['Price'].mean()
print(a)
print(' ')

b = data.groupby('Class')['Price'].std()
print(b)
print(' ')
```

### Ex.06 - Find the prior probability

```python
num = data.groupby('Class')['Price'].count()
print(num)
print(' ')

total = num['fruit'] + num['vegetable']
P_f = num['fruit'] / total
P_v = num['vegetable'] / total

print('Prior probability of fruit = ', P_f)
print('Prior probability of vegetable = ', P_v)
```
