### Ex.02 - Plot histogram

```python
from scipy.stats import uniform

a = 0
b = 1

x = a + (b - a)*np.random.rand(10000)
bin_edges = np.arange(a, b + 0.2, 0.2)

fig, ax = plt.subplots() 
ax.hist(x, bins = bin_edges, density = True)
ax.plot(bin_edges, uniform.pdf(bin_edges, loc = a, scale = (b - a)), 'r-', lw = 4)

ax.grid()
plt.xlabel('x')
plt.ylabel('pdf')
plt.show()
```

### Ex.03 - Binary channel

```python
p_error = 0.2

input_bitstream = [1, 1, 0]

output_bitstream = []
        
for b in input_bitstream:
    print('input bit = ', b)
    if b == 1:
        bit = np.random.rand()
        print('random value = ', bit)
        output_bitstream.append( 1 * (bit >= p_error) )
    else:
        bit = np.random.rand()
        print('random value = ', bit)
        output_bitstream.append( 1 * (bit < p_error) )
    print(output_bitstream)
    print(' ')
```

### Ex.05 - Repetition code

```python
data = [1, 0, 0, 1, 1]
code_length = 3

# Encoding
transmitted = []
for b in data: 
    transmitted.extend([b] * code_length)

print('Transmitted bit sequence = ', transmitted)
print(' ')

# Decoding
received = np.multiply(transmitted, (np.random.rand(len(transmitted)) > 0.5))  # with error
print('Received bit sequence = ', received)
print(' ')

decoded = []
for i in range(0, len(received), code_length):
    decoded.append( 1 * (sum(received[i:i + code_length]) > code_length/2.0) )

print('Original bit sequence = ', data)
print('Decoded bit sequence = ', decoded)
```

### Ex.07 - Useful Probability Distribution

```python
import scipy.stats
from scipy.stats import binom
from scipy.stats import geom
from scipy.stats import poisson
```

```python
n = 10   # number of trials
p = 0.4  # probability of success 
mu = 6   # mu (also often called lambda) parameter of the Poisson distribution

x = np.arange(0, 11)

# pmf
pmfs = np.zeros((3, n+1))
pmfs[0,:] = binom.pmf(x, n=n, p=p)
pmfs[1,:] = geom.pmf(x, p=p)
pmfs[2,:] = poisson.pmf(x, mu=mu)

fig, ax = plt.subplots() 
labels = ('binomial','geometric','poisson')
colors = ('r','g','b')

for i in range(0,3):
    ax.plot(x, pmfs[i,:], 'o-', label = labels[i], color = colors[i])

ax.legend()
ax.grid()
plt.xlabel('x')
plt.ylabel('pmf')
plt.show()
```
