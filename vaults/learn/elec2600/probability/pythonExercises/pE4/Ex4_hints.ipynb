### Ex.05 - Observe the threshold using pdf

```python
x = np.arange(-10, 30, step = 5e-2)

n_pdf1 = scipy.stats.norm.pdf(x, loc = 5, scale = 3) 
n_pdf2 = scipy.stats.norm.pdf(x, loc = 15, scale = 3) 

fig, ax = plt.subplots()

ax.plot(x, n_pdf1)
ax.plot(x, n_pdf2)

plt.grid()
plt.xlabel('x')
plt.ylabel('pdf')
plt.show()
```

### Ex.06 - cdf (B)

```python
x = np.arange(-10, 30, step = 5e-2)

n_pdf2 = scipy.stats.norm.pdf(x, loc = 15, scale = 3) 
n_cdf2 = scipy.stats.norm.cdf(x, loc = 15, scale = 3) 

plt.subplot(211).plot(x, n_pdf2,'orange')
plt.grid()
plt.ylabel('pdf')

plt.subplot(212).plot(x, n_cdf2,'orange')
plt.grid()
plt.ylabel('cdf')
plt.xlabel('x (price)')

plt.show()
```

### Ex.06 - cdf (A)

```python
x = np.arange(-10, 30, step = 5e-2)

n_pdf1 = scipy.stats.norm.pdf(x, loc = 5, scale = 3) 
n_cdf1 = 1 - scipy.stats.norm.cdf(x, loc = 5, scale = 3) 

n_pdf2 = scipy.stats.norm.pdf(x, loc = 15, scale = 3) 
n_cdf2 = scipy.stats.norm.cdf(x, loc = 15, scale = 3) 

plt.subplot(211).plot(x, n_pdf1)
plt.grid()
plt.ylabel('pdf')

plt.subplot(212).plot(x, n_cdf1)
plt.grid()
plt.ylabel('cdf')
plt.xlabel('x (price)')

plt.show()
```

### Ex.07 - Total error probability

```python
x = np.arange(-10, 30, step = 5e-2)

n_cdf1 = scipy.stats.norm.sf(x, loc = 5, scale = 3) 
n_cdf2 = scipy.stats.norm.cdf(x, loc = 15, scale = 3) 

total_cdf = 0.5 * n_cdf1 + 0.5 * n_cdf2
plt.plot(x, total_cdf)

plt.grid()
plt.ylabel('cdf')
plt.xlabel('x (price)')

plt.show()
```

### Ex.08 - Find the min. error probability and optimal threshold

```python
print('Minimum error probability = ', total_cdf.min())
a = total_cdf.argmin()
print('Optimal threshold (Index) = ', a)
print('Optimal threshold (Value) = ', x[a])
```
